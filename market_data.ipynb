```python
import yfinance as yf
import pandas as pd

def fetch_market_dashboard():
    tickers = {
        "S&P 500": "^GSPC",
        "VIX (변동성 지수)": "^VIX",
        "US10Y (미 10년물 국채)": "^TNX",
        "USD/KRW (환율)": "KRW=X",
        "BTC/USD (비트코인)": "BTC-USD"
    }

    print("Market Dashboard\n")

    # 2. 데이터 크롤링 및 포맷팅
    for name, ticker in tickers.items():
        try:
            # 최근 5일치 데이터를 가져와서 전일 종가와 오늘 종가 비교 (주말 처리용)
            data = yf.Ticker(ticker).history(period="5d")
            if len(data) >= 2:
                current_price = data['Close'].iloc[-1]
                prev_price = data['Close'].iloc[-2]

                # 변화량 및 변화율 계산
                diff = current_price - prev_price
                change_pct = (diff / prev_price) * 100

                # 3. 항목별 맞춤형 출력 포맷 (우종님 템플릿에 완벽하게 맞춤)
                if ticker == "^GSPC":
                    print(f"- {name}: {current_price:,.2f}pt (전일대비 {change_pct:+.2f}%)")
                elif ticker == "^VIX":
                    print(f"- {name}: {current_price:.2f} (전일대비 {change_pct:+.2f}%)")
                elif ticker == "^TNX":
                    print(f"- {name}: {current_price:.2f}% (전일대비 {diff:+.2f}%p) 글로벌 역외환율 NDF 기준") # 국채는 %p 단위 사용
                elif ticker == "KRW=X":
                    print(f"- {name}: {current_price:,.0f}원 (전일대비 {diff:+,.0f}원)")
                elif ticker == "BTC-USD":
                    print(f"- {name}: ${current_price:,.0f} (전일대비 {change_pct:+.2f}%)")
        except Exception as e:
            print(f"- {name}: 데이터 로드 실패 ({e})")

if __name__ == "__main__":
    fetch_market_dashboard()
```
